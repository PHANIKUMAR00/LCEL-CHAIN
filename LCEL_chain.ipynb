Objective : To build a simple LCEL cell

```python
#Import the required libraries
import os    #to access environmental variables
from dotenv import load_dotenv  #to load the API keys
from langchain_core.prompts import ChatPromptTemplate  #to create resuable prompts
from langchain_core.output_parsers import StrOutputParser  #converts LLM o/p to simple string
from langchain_groq import ChatGroq    #allows langchain to connect with groq
```

Load the API key

```python
load_dotenv()
api_key=os.getenv("GROQ_API_KEY")
print("API key loaded:",api_key is not None)  #"api_key is not None" checks whether is true or false act like a if else
```

Create the LLM 

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
print("LLM is created Successfully....!")
#temperature is for randomness of LLM
```

```python
response = llm.invoke(
    "Hello"
)
print(response)
```

```python
#Resuable Prompt Template
prompt = ChatPromptTemplate.from_template(
    """"
    Summarize the following text in a simple language
    Text:
    {text}"""
)
```

```python
#Test the Prompt
formatted_prompt = prompt.invoke({
    "text":"Heyy Beauty!"
})
print(formatted_prompt)
```

Build LCEL Chain

```python
#Build LCEL chain
#"|" -->PIPE
chain = prompt|llm
```

```python
parser = StrOutputParser()
chain = prompt|llm|parser
```

```python
results = chain.invoke(
    {
        "text":"write is lang chain"
    }
)
print(results)
```
